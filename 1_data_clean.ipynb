```r
# ====================================
# 1_data_clean.ipynb
# Project: algae_populations_project
# Purpose: To combine raw algal observation data and raw taxa data files into a
#          clean file ready for analysis.
# Reads: raw_data/765141_v6_cpr-list-taxa.csv
#        raw_data/765141_v6_cpr-plankton-abundance.csv
# Writes: clean_data/algae_abundance_clean.csv
# ====================================

setwd("/content/drive/MyDrive/algae_populations_project/clean_data")

library(readr)
library(dplyr)
library(tidyr)
library(lubridate)
```

```r
# Read raw data
taxa_raw <- read.csv("765141_v6_cpr-list-taxa.csv")
abundance_raw <- read.csv("765141_v6_cpr-plankton-abundance.csv")
taxa_raw
abundance_raw
```

```r
# Create a new column for date and time for simplification
abundance_clean <- abundance_raw |>
  mutate(
    month = as.numeric(Month),
    date = make_date(Year, Month, Day),
    datetime = date + hours(Hour)
  )

head(abundance_clean)
```

```r
# Want to identify rows and columns of clean data
glimpse(abundance_clean)
```

```r
# Reshape wide data to long to organize rows by observation
# Drop NA values to remove potential zeros
# Eliminate repition of dates and time
abundance_clean_long <- abundance_clean |>
  pivot_longer(
    cols = starts_with("id_"),
    names_to = "Taxa",
    values_to = "Abundance") |>
  drop_na() |>
  select(-Year, -Month, -Day, -Hour, -month, -date)

head(abundance_clean_long)
```

```r
# Condense raw taxa data to understand corresponding ids
taxa_clean <- taxa_raw |>
  select(Accepted_ID, Taxon_Name) |>
  distinct()

head(taxa_clean)
```

```r
# Convert Accepted_ID into character type
taxa_clean <- taxa_clean |>
  mutate(Accepted_ID = as.character(Accepted_ID))

# Combine data sets to align taxa id with abundance
algae_populations <- abundance_clean_long |>
  left_join(taxa_clean, by = c("Taxa" = "Accepted_ID"))

head(algae_populations)

# REMINDER to fix IDs so taxon is included in dataset
```

```r
# Write clean data
write_csv(algae_populations, "algae_populations.csv")
list.files("clean_data")
```
